```python
# create a list
marks =[98,67,97,99,92,95]
print(marks)
```

```python
student_info=[101,"Harshit","CSE",7.8]
print(student_info)
```

```python
# access the element using index
name=["Harshit","Chirag","devi",'sneha','Ram']
print(name)
```

```python
# access the element using index
name=["Harshit","Chirag","devi",'sneha','Ram']
print(name)
print(name[2])
print(name[1])
print(name[-1])
```

```python
# example for mutable in the list
cgpa=[8.9,9.4,7.4,8.1]
print(cgpa)
```

```python
# example for mutable in the list
cgpa=[8.9,9.4,7.4,8.1]
print(cgpa)
cgpa[2]=7.7
print("After updation")
print(cgpa)
```

```python
# add an iteam in the list
Name=["Kiran","Anil","Ram","Chirag"]
Name.append("Deepika")
print(Name)
```

```python
# insert in particular index
student_name=["Arav","Kapil","Sanjeev"]
student_name.insert(1,"Harsh")
print(student_name)
```

```python
# extend function
sec_a=["Arav","Kapil","Sanjeev"]
sec_b=("Harsh","Ram")
sec_a.extend(sec_b)
print(sec_a)
print(sec_b)
```

```python
# remove an item in the list
sec_a_students=["Arav","Kapil","Sanjeev","Harshit","Aditi"]
sec_a_students.remove("Kapil")
print(sec_a_students)
del sec_a_students[1]
print(sec_a_students)
sec_a_students.pop()
print(sec_a_students)
```

```python
# List slicing
subjects=["Pythons",
          "statistics",
          "SQL",
          "Java","Os",
          "Dsv"]
print(subjects[1:4])
print(subjects[:3])
print(subjects[4:])
print(subjects[::2])
```
